| Позначення     | Пояснення                             |
| -------------- | ------------------------------------- |
| n_A, n_B       | кількість користувачів у групах A і B |
| x_A, x_B       | кількість успіхів (купили, клікнули)  |
| p_A, p_B       | конверсія в групах                    |
| mean_A, mean_B | середнє значення метрики              |
| std_A, std_B   | стандартне відхилення                 |
| delta          | різниця між B і A                     |
| alpha          | рівень значущості (зазвичай 0.05)     |

---

#### z-test (для КОНВЕРСІЙ, 0/1)

```
p = x / n
```

---

#### Різниця конверсій

```
delta = p_B - p_A
```

---

#### Об’єднана (pooled) конверсія

```
p_pooled = (x_A + x_B) / (n_A + n_B)
```

---

#### Стандартна похибка (SE)

```
SE = sqrt( p_pooled * (1 - p_pooled) * (1/n_A + 1/n_B) )
```

---

#### z-статистика

```
z = delta / SE
```

---

#### p-value (двосторонній тест)

```
p_value = 2 * (1 - NormalCDF( |z| ))
```

---

### 95% довірчий інтервал для різниці конверсій
```
SE_diff = sqrt(
    p_A * (1 - p_A) / n_A +
    p_B * (1 - p_B) / n_B
)

CI = delta ± 1.96 * SE_diff
```

---

#### t-test (для ЧИСЛОВИХ МЕТРИК)

#### Різниця середніх

```
delta = mean_B - mean_A
```

---

#### Стандартна похибка (Welch t-test)

```
SE = sqrt( (std_A^2 / n_A) + (std_B^2 / n_B) )
```

---

#### t-статистика

```
t = delta / SE
```

---

### 95% довірчий інтервал для різниці середніх

```
CI = delta ± t_critical * SE
```

(де t_critical ≈ 2 при великих вибірках)

---

#### Інтерпретація z та t

| Значення | Що означає    |
| -------- | ------------- |
| ≈ 0      | різниці немає |
| ≈ 1      | слабкий ефект |
| ≈ 2      | на межі       |
| ≥ 2      | значущий      |
| ≥ 3      | сильний ефект |

---

#### Інтерпретація p-value

| p-value  | Значення            |
| -------- | ------------------- |
| < 0.01   | дуже сильний доказ  |
| < 0.05   | статистично значуще |
| 0.05–0.1 | сумнівно            |
| ≥ 0.1    | доказів немає       |

❗ p-value **НЕ показує розмір ефекту**

---

#### Інтерпретація довірчого інтервалу

| CI        | Висновок             |
| --------- | -------------------- |
| весь > 0  | ефект позитивний     |
| весь < 0  | ефект негативний     |
| містить 0 | результат незначущий |
| вузький   | оцінка точна         |
| широкий   | мало даних / шум     |

---

#### Гіпотези (шаблон)

##### Двосторонній тест

```
H0: A = B
H1: A ≠ B
```

##### Односторонній тест

```
H0: B ≤ A
H1: B > A
```

---

### z-test вручну

```python
z = delta / SE
p_value = 2 * (1 - stats.norm.cdf(abs(z)))
```

---

### t-test (Welch)

```python
from scipy import stats
stats.ttest_ind(A, B, equal_var=False)
```

---

| Ситуація         | Що робити                         |
| ---------------- | --------------------------------- |
| p-value < 0.05   | ефект не випадковий               |
| CI містить 0     | рішення ризиковане                |
| ефект малий      | дивимось на гроші                 |
| незначущий тест  | або мало даних, або слабкий ефект |
| велика дисперсія | t-test може бути ненадійним       |



````python
# Task 
# Порахувати середнє у групі A і B
# Застосувати Welch t-test
# Подивитися на p-value
# Зробити висновок

import numpy as np
from scipy import stats

# Дані
A = np.array([10, 12, 11, 9, 10])
B = np.array([13, 14, 12, 15, 14])

# Крок 1: середні значення
mean_A = A.mean()
mean_B = B.mean()

print("Середній чек A:", mean_A)
print("Середній чек B:", mean_B)

# Крок 2: t-test
t_stat, p_value = stats.ttest_ind(A, B, equal_var=False)
print(t_stat, p_value) # Різниця між групами статистично значуща

# t > 0 - A mean > B mean
# t < 0 - A mean < B mean
````

````python
# Крок 1. Рахуємо конверсії та дельта(pB​−pA)
# Крок 2. Об’єднана конверсія (p = (pa + pb) / count A + count B)
# Крок 3. Стандартна похибка(SE)
# Крок 4. z-статистика дельта / SE
# Крок 5. p-value (двосторонній)
````

````python

import numpy as np
from scipy import stats

nA, xA = 8000, 720
nB, xB = 8000, 784

# Крок 1. Рахуємо конверсії 
pA = xA / nA
pB = xB / nB
print(pA, pB)

# дельта(pB​−pA)
delta = pB - pA
print(delta)

# Крок 2. Об’єднана конверсія (p = (pa + pb) / count A + count B)
p_pool = (xA + xB) / (nA + nB)

# Крок 3. Стандартна похибка(SE)
se = np.sqrt(p_pool * (1 - p_pool) * (1/nA + 1/nB))
print(se)

# Крок 4. z-статистика дельта / SE
z = delta / se

# Крок 5. p-value (двосторонній)
p_value = 2 * (1 - stats.norm.cdf(abs(z))) # cumulative distribution function

z, p_value
````

Task 2. Ви тестуєте новий порядок тарифів.

Метрика: оплата.

Дані:
- A: 12 000 користувачів, 1 020 оплат
- B: 12 000 користувачів, 1 128 оплат

Проведіть z-test (двосторонній, 𝛼=0.05 α=0.05)
та зробіть висновок. Додатково порахуйте CI

````python
nA, xA = 12000, 1020
nB, xB = 12000, 1128

# Крок 1. Конверсії
pA = xA / nA
pB = xB / nB

# дельта(pB​−pA)
delta = pB - pA

# Крок 2. Об’єднана конверсія (p = (pa + pb) / count A + count B)
p_pool = (xA + xB) / (nA + nB) 

# Крок 3. Стандартна похибка
se = np.sqrt(p_pool * (1 - p_pool) * (1/nA + 1/nB))

# Крок 4. z-статистика
z = delta / se

# Крок 5. p-value (двосторонній)
p_value = 2 * (1 - stats.norm.cdf(abs(z))) 

print(z, p_value)

# Крок 6. Порахувати CI
# Довірчий інтервал (Confidence Interval)
# Діапазон значень, у якому з високою ймовірністю лежить реальний ефект

# 95% довірчий інтервал для різниці конверсій

SE_diff = np.sqrt(
    pA * (1 - pA) / nA +
    pB * (1 - pB) / nB
)
CI_p = (delta + 1.96 * SE_diff) * 100
CI_n = (delta - 1.96 * SE_diff) * 100

print(CI_p, CI_n)

# Висновок
# p_value = 0.01 < 0.05 дані статистично значущі
# CI > 0 результат ефективний
````
