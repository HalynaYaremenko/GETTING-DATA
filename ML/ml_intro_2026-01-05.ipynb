#### Machine Learning — це підхід, при якому:
> ми не задаємо правила вручну \

> і робить прогноз або класифікацію для нових ситуацій

Історичні дані + ціль -> навчена модель -> прогноз

**Типові задачі:**
> churn \
> fraud \
> персоналізація \

> сегментація клієнтів

**Аналітика:**
> 12% користувачів пішли \

> частіше йдуть після 7 днів

**ML:**
> дає список користувачів \

### Матриця фіч
> Ця схема таблиці ясно показує, що інформацію можна представити у вигляді двовимірного числового масиву або матриці, яку ми називатимемо *матрицею фіч*.

> За традицією, ця матриця характеристик часто зберігається у змінній з іменем *x*.

> Матриця фіч передбачається двовимірною з формою **[n_samples, n_features]**, і найчастіше міститься в масиві NumPy або Pandas DataFrame, хоча деякі моделі Scikit-learn також приймають розріджені матриці SciPy.

> Вибірки(рядки) завжди відносяться до окремих об'єктів, що описуються набором даних. 

> Ознаки (стовпці) завжди відносяться до окремих спостережень, які кількісно/якісно описують кожен зразок.
Ознаки зазвичай мають речове значення, але в деяких випадках можуть бути булевими або дискретними.

### Знайомство зі Scikit-Learning
> scikit-learn - проєкт з відкритим вихідним кодом, це означає, що його можна вільно використовувати і поширювати, і будь-яка людина може легко отримати вихідний код, щоб побачити, що відбувається "за лаштунками". 

> Проєкт scikit-learn постійно розвивається і вдосконалюється, і в нього дуже активна спільнота користувачів. Він містить низку сучасних алгоритмів машинного навчання, а також повну документацію щодо кожного алгоритму. scikit-learn - дуже популярний інструмент і найвідоміша пітонівська бібліотека для
машинного навчання. Її широко використовують у промисловості та науці, а в інтернеті є багатий вибір навчальних матеріалів і прикладів програмного коду. scikit-learn чудово працює з низкою інших наукових інструментів Python.

> Розуміння принципів роботи та використання scikit-learn - важливе, але є кілька інших бібліотек, які розширять ваш досвід. scikit-learn базується на двох пітонівських бібліотеках для наукових обчислень NumPy і SciPy. Крім NumPy і SciPy ми будемо використовувати pandas і matplotlib. 


> У машинному навчанні з учителем нам потрібно побудувати модель на навчальних даних, а потім отримати точні прогнози для нових даних, які ще не зустрічалися, і які мають такі самі характеристики, що й використаний нами навчальний набір. Якщо модель може видавати точні прогнози на даних, які раніше не зустрічалися, ми говоримо, що модель має здатність узагальнювати (generalize) результат на тестові дані. Нам необхідно побудувати модель, яка матиме максимальну узагальнювальну здатність.  
> Зазвичай ми будуємо модель таким чином, щоб вона давала точні прогнози на навчальному наборі. Якщо навчальний і тестовий набори мають багато спільного між собою, можна очікувати, що модель буде точною і на тестовому наборі. Однак у деяких випадках цього не відбувається.

##### Unit of Prediction (одиниця прогнозу)
Unit of prediction — це:
> обʼєкт \
> для якого модель робить один прогноз

##### Target (цільова змінна)
Target - це:
> те, що модель має передбачити \
> формалізована бізнес-подія


##### Prediction Horizon (горизонт прогнозу)
Horizon — це:
> проміжок часу \
> між моментом прогнозу \
> і моментом події

##### Action (дія після прогнозу)

Для кожного [unit]
на дату T
передбачити [target]
протягом [horizon]
щоб виконати [action]


Для кожного користувача
щодня
передбачити ймовірність churn
протягом 30 днів
щоб запустити retention-кампані

##### k-Nearest Neighbors (kNN)

**kNN:**
- не навчає явну формулу
- зберігає всі тренувальні точки

**для нового обʼєкта:**
- знаходить k найближчих сусідів
- дивиться, які в них target
- голосує / усереднює

Користувач схожий на цих 5 — з них 4 пішли ->  високий ризик churn

Коли kNN має сенс
- маленькі або середні датасети
- інтуїтивна логіка схожості
- погано масштабується
-  дуже чутливий до масштабу ознак

```python
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV, train_test_split

import pandas as pd
import numpy as np 
import seaborn  as sns
import matplotlib.pyplot as plt

%matplotlib inline 
sns.set('notebook')

import warnings

warnings.filterwarnings("ignore", category=UserWarning)
warnings.filterwarnings("ignore", category=FutureWarning)
```

```python
df = sns.load_dataset("iris")
df.head()
```

```python
df['species'].unique()
```

```python

X = df.drop(columns="species")
y = df["species"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled[0:5], X_train[0:5]
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)

y_pred = knn.predict(X_test_scaled)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
##### Decision Tree
df = sns.load_dataset("titanic")

df = df[["survived", "pclass", "sex", "age", "fare"]]
df = df.dropna()
```

```python
df["sex"] = df["sex"].map({"male": 0, "female": 1})
```

```python
X = df.drop(columns="survived")
y = df["survived"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
```

```python
tree = DecisionTreeClassifier(
    max_depth=4,
    min_samples_leaf=30,
    random_state=42
)

tree.fit(X_train, y_train)

y_pred_proba = tree.predict_proba(X_test)[:, 1]
```

```python
tree.feature_importances_
```

```python
##### Регресія

df = sns.load_dataset("tips")
df.head()
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error

X = df[["total_bill", "size"]]
y = df["tip"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred = lr.predict(X_test)
print("MAE:", mean_absolute_error(y_test, y_pred))
```

```python
for feature, coef in zip(X.columns, lr.coef_):
    print(feature, coef)
```
